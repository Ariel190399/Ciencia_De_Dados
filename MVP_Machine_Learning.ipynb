# **Descrição do Problema**
O dataset Electric Vehicles Specification 2025 contém informações técnicas detalhadas sobre diversos modelos de veículos elétricos disponíveis no mercado global.  
O objetivo principal deste estudo é investigar o impacto das características técnicas como velocidade máxima, capacidade da bateria, torque e eficiência energética no desempenho dos veículos, com ênfase especial na autonomia (autonomia_km).

Através de uma análise exploratória aprofundada, busco identificar padrões, correlações e distribuições desses atributos, a fim de embasar a construção de modelos preditivos robustos para estimar a autonomia dos veículos elétricos com base em suas especificações técnicas.

* Veículos com maior torque (Nm) apresentam maior velocidade máxima (km/h)?



*   Há uma relação negativa entre a eficiência energética (Wh/km) e a autonomia (km)?



*  Veículos com maior velocidade máxima (km/h) tendem a apresentar menor eficiência energética (Wh/km)?



*   A capacidade da bateria (kWh) está positivamente correlacionada com o torque do veículo (Nm)?

# **Atributos do Dataset**
O dataset Electric Vehicles Specification 2025 contém dados técnicos detalhados de diversos modelos de veículos elétricos comercializados globalmente. Para a análise inicial, selecionei as cinco principais variáveis abaixo:



*  Velocidade Máxima (velocidade_max_kmh): A velocidade máxima que o veículo pode atingir, medida em quilômetros por hora (km/h).

*  Eficiência Energética (eficiencia_wh_por_km): O consumo médio de energia elétrica do veículo por quilômetro rodado, medido em watt-hora por quilômetro (Wh/km).

 **MÍNIMO E MÁXIMO**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Selecionar apenas colunas numéricas
df_numeric = df.select_dtypes(include=['number'])

# Calcular mínimo e máximo de todas as colunas numéricas
resumo_min_max = pd.DataFrame({
    'Mínimo': df_numeric.min(),
    'Máximo': df_numeric.max()
})

# Exibir o resumo no console
print("\n✅ Resumo de Mínimos e Máximos de todas as colunas numéricas:\n")
print(resumo_min_max)

# Lista das 5 principais variáveis
variaveis_principais = [
    'velocidade_max_kmh',
    'autonomia_km',
    'capacidade_bateria_kWh',
    'torque_nm',
    'eficiencia_wh_por_km'
]

# Cálculo dos mínimos e máximos das 5 principais variáveis
minimos = df[variaveis_principais].min()
maximos = df[variaveis_principais].max()

# Configuração do eixo x
x = np.arange(len(variaveis_principais))
largura = 0.35  # Largura das barras

# Definindo o limite superior do eixo y com folga
y_limite_superior = max(maximos.values) * 1.15

# Criação do gráfico
plt.figure(figsize=(12, 7))
barras_min = plt.bar(x - largura / 2, minimos.values, largura, label='Mínimo', color='green')
barras_max = plt.bar(x + largura / 2, maximos.values, largura, label='Máximo', color='red')

# Adicionando os valores logo acima das barras, próximos delas
for barra in barras_min:
    altura = barra.get_height()
    plt.text(barra.get_x() + barra.get_width() / 2, altura + (y_limite_superior * 0.01), f'{altura:.1f}',
             ha='center', va='bottom')

for barra in barras_max:
    altura = barra.get_height()
    plt.text(barra.get_x() + barra.get_width() / 2, altura + (y_limite_superior * 0.01), f'{altura:.1f}',
             ha='center', va='bottom')

# Configurações finais do gráfico
plt.ylim(0, y_limite_superior)
plt.xticks(x, variaveis_principais, rotation=45)
plt.ylabel('Valor')
plt.title('Mínimo e Máximo das 5 Principais Variáveis')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

O mínimo e o máximo são métricas estatísticas básicas, mas fundamentais, utilizadas na análise exploratória de dados para entender rapidamente o comportamento de uma variável.





Mínimo: Permite identificar o limite inferior dos dados e possíveis valores fora do esperado (ex: veículos com a menor autonomia ou menor torque).

Máximo: Indica o limite superior dos dados, ajudando a perceber o melhor desempenho, maior capacidade ou casos extremos.





*  **Velocidade Máxima:** De 125 km/h a 325 km/h, revelando a presença de modelos básicos urbanos e veículos esportivos de alto desempenho.

*  **Autonomia:** Varia de 135 km a 685 km, abrangendo desde carros para o dia a dia até modelos com foco em longas distâncias.

*   **Capacidade da Bateria:** Entre 21.3 kWh e 118 kWh, o que influencia diretamente a autonomia e o perfil do veículo (menores baterias tendem a estar em modelos compactos).

*  **Torque:** Oscila de 113 Nm a 1350 Nm, demonstrando que há tanto veículos de perfil econômico quanto modelos mais robustos e potentes.


*  **Eficiência:** De 109 Wh/km a 370 Wh/km, onde veículos mais eficientes consomem menos energia por quilômetro, enquanto modelos maiores ou mais potentes possuem consumo elevado.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Selecionar apenas colunas numéricas
df_numeric = df.select_dtypes(include=['number'])

# Exibir todas as modas no console
print("✅ Modas de todas as variáveis numéricas do DataFrame:\n")
for coluna in df_numeric.columns:
    modas = df[coluna].mode()
    modas_str = ', '.join([f"{m:.1f}" for m in modas])
    print(f"{coluna}: {modas_str}")

# Garantir que a coluna velocidade_max_kmh está no formato numérico
df['velocidade_max_kmh'] = pd.to_numeric(df['velocidade_max_kmh'], errors='coerce')

# Calcular as modas da velocidade máxima
modas_velocidade = df['velocidade_max_kmh'].mode()
primeira_moda = modas_velocidade.iloc[0]

# Gerar o gráfico KDE da velocidade máxima
plt.figure(figsize=(10, 6))
sns.kdeplot(df['velocidade_max_kmh'].dropna(), color='green', fill=True)

# Adicionar apenas a primeira moda no gráfico
plt.axvline(primeira_moda, color='red', linestyle='-', label=f'Moda: {primeira_moda:.1f} km/h')

# Ajustes finais do gráfico
plt.title('Densidade - Velocidade Máxima')
plt.xlabel('Velocidade Máxima (km/h)')
plt.ylabel('Densidade')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

A moda é a medida de tendência central que indica o valor mais frequente em um conjunto de dados, ou seja, o valor que mais se repete.


O valor mais frequente de velocidade máxima encontrado no dataset é de aproximadamente 180 km/h.

Esse valor está destacado no gráfico de densidade com uma linha vermelha, indicando que muitos veículos elétricos do conjunto de dados possuem essa velocidade máxima.





*   **Moda:**  180 km/h

*  **Variação:** Presença de modelos mais básicos e outros de alta performance, mas menos frequentes

**DISTRIBUIÇÃO INDIVIDUAL DAS VARIÁVEIS.**

**TRATAMENTO DE VALORES FALTANTES**

```python
# Mostrar valores ausentes por coluna
print(df.isnull().sum())

# Exibir proporção de valores faltantes para entender impacto
print((df.isnull().mean() * 100).round(2))

# Imputamos valores ausentes: mediana para numéricas e moda para categóricas.
# Pode-se também remover linhas com muitos nulos.

from sklearn.impute import SimpleImputer

# Separar colunas numéricas e categóricas
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
cat_cols = df.select_dtypes(include=['object', 'category']).columns

# Imputar colunas numéricas
imputer_num = SimpleImputer(strategy='median')
df[num_cols] = imputer_num.fit_transform(df[num_cols])

# Imputar colunas categóricas
imputer_cat = SimpleImputer(strategy='most_frequent')
df[cat_cols] = imputer_cat.fit_transform(df[cat_cols])

# Verificar se ainda há valores nulos
print(df.isnull().sum())
```

O tratamento de valores faltantes consiste em identificar e corrigir registros incompletos ou nulos presentes no dataset. Esse processo é essencial para garantir a confiabilidade das análises, dos cálculos estatísticos e da performance de modelos preditivos.

**Diagnóstico inicial:**

O código começa por identificar exatamente quantos valores ausentes existem em cada coluna e a proporção percentual desses valores em relação ao total. Isso é crucial para entender o impacto dos dados faltantes e decidir estratégias apropriadas para tratá-los.

**Diferenciação de tipos de dados:**

Ele separa as colunas em numéricas e categóricas, reconhecendo que o tratamento para cada tipo deve ser diferente — uma boa prática no pré-processamento.

**Imputação de valores ausentes:**



*   Para colunas numéricas, o código usa a mediana para substituir os valores ausentes. A mediana é uma escolha robusta porque não é influenciada por outliers, preservando melhor a distribuição dos dados.





*   Para colunas categóricas, a substituição é feita pelo valor mais frequente (moda), que é uma maneira razoável de preencher valores faltantes em variáveis qualitativas, mantendo a consistência dos dados.



**Revisão pós-imputação:**
Após o preenchimento, o código verifica novamente a presença de valores ausentes, garantindo que o tratamento foi eficaz e o conjunto de dados está pronto para as próximas etapas.

**NORMALIZAÇÃO**

```python
from sklearn.preprocessing import MinMaxScaler

# Definir as 5 variáveis numéricas principais
variaveis_numericas = [
    'velocidade_max_kmh',
    'autonomia_km',
    'capacidade_bateria_kWh',
    'torque_nm',
    'eficiencia_wh_por_km'
]

# Criar uma cópia do dataframe para preservar o original
df_normalizado = df.copy()

# Inicializar o scaler Min-Max (escala de 0 a 1)
scaler = MinMaxScaler()

# Aplicar a normalização apenas nas colunas escolhidas
df_normalizado[variaveis_numericas] = scaler.fit_transform(df[variaveis_numericas])

# Mostrar resumo estatístico antes e depois para comparação
print("Antes da normalização:\n", df[variaveis_numericas].describe())
print("\nDepois da normalização:\n", df_normalizado[variaveis_numericas].describe())
```

Normalização é uma técnica de pré-processamento que ajusta os valores das variáveis numéricas para uma escala comum, geralmente entre 0 e 1. Isso é feito para que diferentes variáveis, que podem ter unidades e magnitudes muito diferentes, fiquem comparáveis entre si.



*   Antes da normalização, os valores tinham escalas muito diferentes: por exemplo, torque varia de 113 a 1350 Nm, enquanto autonomia vai de 135 a 685 km. Essa disparidade pode prejudicar modelos que usam distância ou gradientes.





*   Após a normalização, todos os valores estão no intervalo [0,1]. Isso deixa todas as variáveis na mesma escala, facilitando o aprendizado do modelo, evitando que variáveis com maior magnitude dominem.



Estatísticas após normalização:



*  As médias e desvios padrão caíram para valores menores (médias entre ~0.2 e ~0.5 e desvios entre ~0.13 e ~0.19), o que é esperado na escala 0-1.



**PADRONIZAÇÃO**

```python
from sklearn.preprocessing import StandardScaler

variaveis_principais = [
    'velocidade_max_kmh',
    'autonomia_km',
    'capacidade_bateria_kWh',
    'torque_nm',
    'eficiencia_wh_por_km'
]

scaler = StandardScaler()

df_scaled = df.copy()
df_scaled[variaveis_principais] = scaler.fit_transform(df[variaveis_principais])

print("Estatísticas após padronização:")
print(df_scaled[variaveis_principais].describe().T[['mean', 'std', 'min', 'max']])
```

Padronização (também chamada de z-score normalization) é uma técnica de pré-processamento que transforma os dados para que tenham média zero e desvio padrão 1.



*  Torna as variáveis comparáveis, mesmo que estejam em escalas diferentes.

*   É essencial para algoritmos que usam distância (KNN, SVM, regressão logística, redes neurais, etc).

*   Ajuda o algoritmo a convergir mais rápido e evita que variáveis com valores maiores dominem o modelo.
















DADOS:

*   Média próxima de zero em todas as variáveis: indica que o centro dos dados está ajustado para zero, como esperado

*  Desvio padrão próximo de 1: as variáveis agora têm dispersão padrão, o que facilita comparações

*   Valores mínimo e máximo variando: representam os outliers e a amplitude dos dados padronizados.


**ONE-HOT ENCODING**

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd


colunas_para_ohe = ['marca', 'segmento_agrupado']

# Criar o OneHotEncoder
# Usaei sparse_output=False para evitar erro nas versões mais recentes
ohe = OneHotEncoder(drop='first', sparse_output=False)

# Ajustar e transformar os dados
onehot_array = ohe.fit_transform(df[colunas_para_ohe])

# Obter nomes das colunas geradas
colunas_ohe = ohe.get_feature_names_out(colunas_para_ohe)

# Criar DataFrame com as colunas one-hot
df_ohe = pd.DataFrame(onehot_array, columns=colunas_ohe, index=df.index)

# Juntar com o DataFrame original, removendo as colunas originais
df_final = pd.concat([df.drop(columns=colunas_para_ohe), df_ohe], axis=1)

print(df_final.head())
```

One-Hot Encoding é uma técnica de transformação de variáveis categóricas em variáveis numéricas, criando uma coluna binária (0 ou 1) para cada categoria distinta. Isso permite que algoritmos de machine learning, que geralmente trabalham apenas com dados numéricos, processem informações categóricas.

Por exemplo, uma variável segmento com categorias Compact, Luxury e Medium será transformada em três colunas:


*   segmento_Compact

*   segmento_Luxury

*  segmento_Medium

Cada linha terá valor 1 na coluna correspondente à sua categoria e 0 nas demais.

DADOS:

No meu código, apliquei o One-Hot Encoding para as variáveis categóricas do meu dataset, como o segmento e outras colunas que agrupei previamente. Antes disso, fiz um agrupamento das categorias menos frequentes em uma categoria única chamada “Outros”, o que reduziu o número total de categorias e evitou que o conjunto final tivesse muitas colunas esparsas e pouco representativas.

Como resultado, criei várias colunas binárias uma para cada categoria aumentando o número total de colunas do dataset, mas tornando-o adequado para a aplicação de modelos de machine learning.

Essa transformação é fundamental para modelos que não trabalham diretamente com dados categóricos e me ajudou a preservar a informação sem impor nenhuma ordem ou hierarquia entre as categorias.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import r2_score

# Split
X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y, test_size=0.20, random_state=42
)

# Modelos com hiperparâmetros iniciais
modelos = {
    "Ridge": Ridge(alpha=1.0, random_state=42),
    "Lasso": Lasso(alpha=0.1, random_state=42, max_iter=10000),
    "Random Forest": RandomForestRegressor(n_estimators=100, max_depth=None, random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(n_estimators=200, learning_rate=0.1, max_depth=3, random_state=42),
}

# Treino e avaliação
res = []
for nome, m in modelos.items():
    m.fit(X_train, y_train)
    res.append({
        "Modelo": nome,
        "R² Treino": r2_score(y_train, m.predict(X_train)),
        "R² Teste":  r2_score(y_test,  m.predict(X_test)),
    })

pd.DataFrame(res).sort_values("R² Teste", ascending=False)
```

**OTIMIZAÇÃO DE HIPERPARÂMENTROS**

Após o treinamento inicial e análise de underfitting/overfitting, foi realizada a otimização dos hiperparâmetros dos modelos selecionados. Para isso, utilizou-se a técnica de GridSearchCV, que avalia diferentes combinações de parâmetros através de validação cruzada, escolhendo aquela que maximiza a performance no conjunto de treino sem comprometer a generalização.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Ridge, Lasso
from sklearn.ensemble import GradientBoostingRegressor

# ===== Ridge =====
param_grid_ridge = {"alpha": [0.01, 0.1, 1, 10, 100]}
grid_ridge = GridSearchCV(Ridge(random_state=42), param_grid_ridge, cv=5, scoring="r2")
grid_ridge.fit(X_train, y_train)

# ===== Lasso =====
param_grid_lasso = {"alpha": [0.0001, 0.001, 0.01, 0.1, 1]}
grid_lasso = GridSearchCV(Lasso(random_state=42, max_iter=10000), param_grid_lasso, cv=5, scoring="r2")
grid_lasso.fit(X_train, y_train)

# ===== Gradient Boosting =====
param_grid_gb = {
    "n_estimators": [100, 200, 300],
    "learning_rate": [0.01, 0.05, 0.1],
    "max_depth": [3, 5, 7],
}
grid_gb = GridSearchCV(GradientBoostingRegressor(random_state=42), param_grid_gb, cv=5, scoring="r2")
grid_gb.fit(X_train, y_train)

# ===== Resultados =====
print("Melhores parâmetros Ridge:", grid_ridge.best_params_)
print("R² treino:", grid_ridge.score(X_train, y_train))
print("R² teste:", grid_ridge.score(X_test, y_test))

print("\nMelhores parâmetros Lasso:", grid_lasso.best_params_)
print("R² treino:", grid_lasso.score(X_train, y_train))
print("R² teste:", grid_lasso.score(X_test, y_test))

print("\nMelhores parâmetros Gradient Boosting:", grid_gb.best_params_)
print("R² treino:", grid_gb.score(X_train, y_train))
print("R² teste:", grid_gb.score(X_test, y_test))
```

Foi realizada a otimização dos hiperparâmetros utilizando GridSearchCV. Os modelos Ridge, Lasso e Gradient Boosting foram ajustados com diferentes combinações de parâmetros, buscando maximizar o R² e garantir boa capacidade de generalização.

**AVALIAÇÃO DE MÉTODOS AVANÇADOS**

```python
import re
import pandas as pd

def sanitize_name(name: str) -> str:
    s = re.sub(r'[^A-Za-z0-9_]', '_', str(name))  # troca tudo que não for A-z0-9_ por _
    if re.match(r'^\d', s):  # evita começar com número
        s = f'F_{s}'
    s = re.sub(r'_+', '_', s).strip('_')  # colapsa underscores e remove das pontas
    return s or "feature"

def sanitize_columns(df: pd.DataFrame) -> pd.DataFrame:
    new_cols, seen = [], {}
    for c in df.columns:
        base = sanitize_name(c)
        if base in seen:
            seen[base] += 1
            new_cols.append(f"{base}__{seen[base]}")  # garante unicidade
        else:
            seen[base] = 0
            new_cols.append(base)
    out = df.copy()
    out.columns = new_cols
    return out

# Saneia e alinha colunas de treino/teste
X_train_san = sanitize_columns(X_train)
X_test_san  = sanitize_columns(X_test)
X_test_san  = X_test_san.reindex(columns=X_train_san.columns, fill_value=0)
```

Remove caracteres especiais dos nomes das colunas, garante que nenhuma comece com número, evita colisões (adicionando sufixos), e alinhe as colunas do teste às do treino (mesma ordem/mesmo conjunto).

**COMITÊ DE MODELOS (ENSEMBLES)**

```python
from sklearn.linear_model import Ridge
from sklearn.ensemble import GradientBoostingRegressor, VotingRegressor, StackingRegressor
from xgboost import XGBRegressor
from sklearn.metrics import r2_score
import pandas as pd

# ===== Modelos base (com hiperparâmetros finais) =====
ridge_model = Ridge(alpha=0.01, random_state=42)
gb_model = GradientBoostingRegressor(
    n_estimators=300, learning_rate=0.05, max_depth=3, random_state=42
)
xgb_model = XGBRegressor(
    n_estimators=600, learning_rate=0.05, max_depth=6,
    subsample=0.8, colsample_bytree=0.8,
    reg_alpha=0.0, reg_lambda=1.0,
    random_state=42, n_jobs=-1, tree_method="hist"
)

# Treinar os modelos-base
ridge_model.fit(X_train_san, y_train)
gb_model.fit(X_train_san, y_train)
xgb_model.fit(X_train_san, y_train)

# ===== Comitê 1: Voting Regressor =====
voting = VotingRegressor([
    ("ridge", ridge_model),
    ("gb", gb_model),
    ("xgb", xgb_model)
])
voting.fit(X_train_san, y_train)

# ===== Comitê 2: Stacking Regressor =====
stacking = StackingRegressor(
    estimators=[
        ("ridge", ridge_model),
        ("gb", gb_model),
        ("xgb", xgb_model)
    ],
    final_estimator=Ridge(alpha=1.0, random_state=42),
    cv=5,
    n_jobs=-1
)
stacking.fit(X_train_san, y_train)

# ===== Avaliação =====
resultados_ensemble = []
for nome, modelo in [("Voting", voting), ("Stacking", stacking)]:
    r2_tr = r2_score(y_train, modelo.predict(X_train_san))
    r2_ts = r2_score(y_test, modelo.predict(X_test_san))
    resultados_ensemble.append({"Modelo": nome, "R² Treino": r2_tr, "R² Teste": r2_ts})

df_ensemble = pd.DataFrame(resultados_ensemble)
print(df_ensemble)
```

Nesta etapa foram avaliadas técnicas de *ensemble*, combinando diferentes algoritmos para verificar se a integração de modelos aumenta a capacidade de generalização.

#### Modelos escolhidos para o comitê
Com base nos resultados anteriores, foram selecionados os três melhores modelos para compor o comitê:
- **Ridge Regression:** (estabilidade e simplicidade).  
- **Gradient Boosting Regressor:** (bom equilíbrio viés/variância).  
- **XGBoost Regressor:** (melhor desempenho individual).  

####  Estratégias utilizadas
1. **Voting Regressor (média das previsões)**  
   - Combina as previsões de cada modelo e retorna a média.  
   - Boa estratégia quando os modelos têm desempenhos similares, reduzindo variância.  

2. **Stacking Regressor: (meta-modelo)**  
   - Usa um modelo de nível superior (meta-estimador) para aprender como combinar as previsões dos modelos base.  
   - Neste caso, foi utilizado um Ridge como meta-modelo.  

### Resultados obtidos

| Modelo   | R² Treino | R² Teste | Análise |
|----------|-----------|----------|---------|
| **Voting**   | 0.999296 | 0.984065 | Superou os modelos individuais e ficou acima do XGBoost (0.9825). Mostrou maior estabilidade e robustez. |
| **Stacking** | 0.999928 | 0.984549 | Melhor resultado entre todos os testes. O meta-modelo Ridge aprendeu a ponderar as previsões, aumentando a capacidade de generalização. |

### Conclusão
- O Voting Regressor já apresentou melhoria em relação ao melhor modelo individual (XGBoost).  
- O Stacking Regressor foi o destaque, alcançando **R² Teste = 0.9845**, tornando-se o melhor modelo do projeto.  
- Esses resultados confirmam que a combinação de algoritmos pode superar modelos isolados, trazendo ganhos em desempenho e robustez.
